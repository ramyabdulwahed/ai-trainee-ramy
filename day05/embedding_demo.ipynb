```python
#5.2 - sentence transformers hands on

from sentence_transformers import SentenceTransformer, util #SentenceTransformer is the class that loads embedding models. Util = helper functions to compute cosine sim

model = SentenceTransformer ("all-MiniLM-L6-v2") #Download/load this embedding model

sentences = [
    "A man is eating pasta.",
    "Someone is enjoying a bowl of spaghetti.",
    "The stock market dropped today.",
    "A woman plays the violin on stage.",
]

emb = model.encode ( #turn each sentence into an embedding vector
    sentences,
    convert_to_tensor=True,
    normalize_embeddings=True #make each sentence vector have L2 length 1
)

print (emb.shape) #Read that as: 4 sentence embeddings 384 numbers per embedding

sims = util.cos_sim (emb, emb) #This compares: every sentence embedding against every sentence embedding

print (sims.round(decimals=3)) #round to 3 decimal places. we get a tensor back
```

```python
#script to test between 3 models
from sentence_transformers import SentenceTransformer, util

models_to_test = [
    "all-MiniLM-L6-v2",
    "BAAI/bge-small-en-v1.5",
    "sentence-transformers/all-mpnet-base-v2",
]

test_cases = [
    {
        "query": "A child is playing soccer.",
        "candidates": [
            "A kid is kicking a football in the park.",
            "A chef is preparing soup.",
            "The laptop battery is almost dead.",
            "A singer performed on stage.",
        ],
        "correct_index": 0,
    },
    {
        "query": "My computer will not charge.",
        "candidates": [
            "The phone camera takes clear pictures.",
            "The laptop battery is almost dead.",
            "A team won the football match.",
            "Someone is cooking dinner.",
        ],
        "correct_index": 1,
    },
    {
        "query": "A person is making food.",
        "candidates": [
            "The stock market increased yesterday.",
            "A chef is preparing soup.",
            "A musician is tuning a violin.",
            "The car needs new tires.",
        ],
        "correct_index": 1,
    },
]

for model_name in models_to_test:
    print("\n" + "=" * 80)
    print("MODEL:", model_name)
    print("=" * 80)

    model = SentenceTransformer(model_name)

    correct = 0

    for case in test_cases:
        query = case["query"]
        candidates = case["candidates"]
        correct_index = case["correct_index"]

        candidate_emb = model.encode(
            candidates,
            convert_to_tensor=True,
            normalize_embeddings=True
        )

        query_emb = model.encode(
            query,
            convert_to_tensor=True,
            normalize_embeddings=True
        )

        scores = util.cos_sim(query_emb, candidate_emb)[0]
        best_index = int(scores.argmax())

        is_correct = best_index == correct_index

        if is_correct:
            correct += 1

        print("\nQuery:", query)
        print("Best match:", candidates[best_index])
        print("Expected:", candidates[correct_index])
        print("Correct?", is_correct)

    accuracy = correct / len(test_cases)
    print(f"\nTiny accuracy: {accuracy:.2f}")
```
