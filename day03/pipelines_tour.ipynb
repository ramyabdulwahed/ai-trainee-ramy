```python
#sentiment analysis
from transformers import pipeline
#Hugging Face is hiding three steps: 
# Step 1: Tokenizer: The model does not understand English words directly. So first, the tokenizer converts text into numbers.
# Step 2: Model The model takes those numbers and does calculations. It produces raw scores called logits. A logit is just the models raw, unclean final score before converting it into a readable probability.
# Step 3: Labels Then Hugging Face converts those raw scores into a readable answer: POSITIVE, 0.999 That final readable result is the label and confidence score.
sa = pipeline ("sentiment-analysis") #Give me a ready to use pipeline for the task of sentiment analysis
sa(["I loved this movie.", "Waste of two hours.", "It was... fine."]) #Sa now holds the model object full workflow. ex.tokenizer,model,output
```

```python
#zero shot classification
#zero-shot classification means: Classify text into labels we choose, without training the model on specific labels. So we can give it a sentence and invent labels on the spot.
zsc = pipeline("zero-shot-classification", model="facebook/bart-large-mnli") #MNLI is a dataset/task where the model learns whether one sentence supports another sentence, this trick is used by zero shot classification
zsc ("TReal Madrid signed a new young midfielder after an impressive season.", candidate_labels=["economics", "sports", "technology", "entertainment"])
```

```python
#name entity recognition: finds important named things inside a text. info extraction
ner = pipeline ("ner",  model="dslim/bert-base-NER", aggregation_strategy="simple")
ner ("Barack Obama was born in Hawaii and graduated from Harvard.")
```

```python
#question answering

qa = pipeline ("question-answering", model="distilbert-base-cased-distilled-squad")
qa (question="Where did Obama graduate?",
    context="Barack Obama was born in Hawaii and graduated from Harvard.") #Extractive QA means: The answer must be copied from the given context. The model is not freely generating an answer like Chat/copilot
```

```python
#summarization:takes a longer text and produce a shorter version
summ = pipeline ("summarization",  model="sshleifer/distilbart-cnn-12-6")
article = """
Hugging Face has become one of the most widely used platforms for sharing machine learning models,
datasets, and demos. Developers can download pretrained models, fine-tune them on their own data,
and deploy them into applications. The platform also includes tools such as Transformers, Datasets,
and Spaces, which make it easier to build and test AI systems quickly.
"""
summ (article,  max_length=40, min_length=15)
```

```python
#Step 3.3 — Tokenizers Deep Dive
#Before this, we used pipeline() as the easy button. Now we open the box a bit and ask: What actually happens to my sentence before it reaches the model? 
# The answer: Text must be converted into numbers. Because models do not understand raw English directly. They understand numbers.

from transformers import AutoTokenizer #A tokenizer is the thing that converts text into model-readable numbers. auto will load the correct tokenzier for the model name we give to hugging face
tok = AutoTokenizer.from_pretrained ("bert-base-uncased") #tok holds the converter
enc = tok ("Hello, World!",return_tensors="pt") #pt = return output as pytorch tensors, ex. number containers used by the deep learning lib
print (enc)

#ouptut {'input_ids': tensor([[ 101, 7592, 1010, 2088,  999,  102]]), 'token_type_ids': tensor([[0, 0, 0, 0, 0, 0]]), 'attention_mask': tensor([[1, 1, 1, 1, 1, 1]])}
#input_ids = the actual token numbers
#attention_mask = real token or padding?
```

```python
#Step 3.3 Tokenizers Deep Dive - Word splitting

words = ["playing",
    "tokenization",
    "unbelievable",
    "Ramy",
    "Guelph"]

for word in words:
    ids = tok (word)["input_ids"] #convert words into tokens
    tokens = tok.convert_ids_to_tokens(ids) #converts tokens into words
    print (word, "->", tokens)
#output
# playing -> ['[CLS]', 'playing', '[SEP]']
# tokenization -> ['[CLS]', 'token', '##ization', '[SEP]']
# unbelievable -> ['[CLS]', 'unbelievable', '[SEP]']
# Ramy -> ['[CLS]', 'ram', '##y', '[SEP]']
# Guelph -> ['[CLS]', 'gu', '##el', '##ph', '[SEP]']

#output explained:
#What does ## mean? In BERT WordPiece tokenization: ##ization means: This is not a full standalone word. It is attached to the previous token. 
# So: token + ##ization = tokenization The model vocabulary may not contain every possible word, so it breaks uncommon words into smaller known pieces.
#This is useful because the model can handle words it has never seen exactly before.
```

```python
#Step 3.3 Tokenizers Deep Dive - uncased vs cased

#holds tokenizers
uncased_tok = AutoTokenizer.from_pretrained ("bert-base-uncased") 
cased_tok = AutoTokenizer.from_pretrained ("bert-base-cased")

text = "Apple released a new product. I ate an apple."
print ("unbiased")
print (uncased_tok.convert_ids_to_tokens (uncased_tok (text)["input_ids"])) #we convert words to tokens and then back to words

print("Cased:")
print(cased_tok.convert_ids_to_tokens(cased_tok(text)["input_ids"]))
```

```python
#Step 3.3 Tokenizers Deep Dive - padding 
#What is padding? The two sentences have different lengths: Good movie. is short. This movie was surprisingly emotional and beautifully written. is longer. But tensors like rectangle shapes. 
# So the tokenizer makes both sequences the same length by adding [PAD] tokens to the shorter one.
sentences = ["Good movie.",
             "This movie was surprisingly emotional and beautifully written." ]

batch = tok (sentences, padding=True, truncation=True, return_tensors="pt")

print(batch)
print(tok.convert_ids_to_tokens(batch["input_ids"][0]))
print(tok.convert_ids_to_tokens(batch["input_ids"][1]))
print(batch["attention_mask"]) #tells the model whether whether a token is real or padding



#function defined
def tokenize_batch (text):
    return tok (text, padding=True, truncation=True, max_length=256, return_tensors="pt")
```
