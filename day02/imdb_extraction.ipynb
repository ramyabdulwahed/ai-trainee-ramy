```python
!pip install datasets -q
from datasets import load_dataset ##huggice face dataset loader to download public dataset without kaggle
import pandas as pd

ds = load_dataset("stanfordnlp/imdb") ##ds holds a hugging face object with the training and testing data
```

```python
print (ds)
```

```python
#now we convert the dataset to a csv file so we can use it without huggingface
#take hugging face train and test data and put it into pandas tables respectively
df_train = ds["train"].to_pandas()
df_test = ds["test"].to_pandas()

label_map = {0: "negative",
             1 : "positive"}

df_train["sentiment"] = df_train["label"].map(label_map) #instead of seeing 0s and 1s we see neg and pos
df_test["sentiment"] = df_test["label"].map(label_map)


#save files now
df_train.to_csv ("imdb_train.csv", index=False)
df_test.to_csv ("imdb_test.csv", index=False)
```
